```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
```

```python
df_titanic = sns.load_dataset("titanic")
df_titanic.head()
```

```python
sns.relplot(data=df_titanic, x="class", y="age", hue="alive")
```

```python
titanic=sns.load_dataset("titanic")
titanic
sns.boxplot(x=titanic["age"])
```
